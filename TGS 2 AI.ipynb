##  MENENTUKAN PARAMETER GEMPA
Dalam bidang geofisika, salah satu permasalahan yang sering dihadapi oleh seorang geofisikawan adalah mengidentifikasi parameter gempa bumi berdasarkan data waktu tiba gelombang P dan gelombang S. Data tersebut dapat digunakan untuk menentukan waktu terjadinya gempa atau *origin time* serta memperkirakan parameter lain yang berkaitan dengan sumber gempa.

Salah satu metode yang dapat digunakan untuk menganalisis hubungan antara waktu tiba gelombang P dan S adalah **Diagram Wadati**. Metode ini memanfaatkan perbedaan waktu tiba gelombang S dan P dari beberapa stasiun seismik. Dengan melakukan analisis terhadap hubungan antara waktu tiba gelombang tersebut, dapat diperoleh informasi mengenai **origin time (T₀)** dan rasio kecepatan gelombang **Vp/Vs**.

Dalam praktiknya, data seismik yang diperoleh dari beberapa stasiun dapat memiliki perbedaan waktu tiba gelombang P dan S. Oleh karena itu, diperlukan proses pengolahan data yang sistematis untuk mendapatkan parameter tersebut dan memastikan hasil analisis sesuai dengan prinsip dasar seismologi.

Pada tugas ini, kecerdasan artifisial digunakan untuk membantu seorang geofisikawan dalam mengolah dan menganalisis data waktu tiba gelombang P dan S menggunakan metode Diagram Wadati. AI tidak hanya digunakan untuk memberikan hasil akhir, tetapi juga membantu menjelaskan proses pengolahan dan perhitungan secara bertahap sehingga hasil analisis dapat dipahami dan diverifikasi.

# RENCANA HASIL AKHIR

Berdasarkan struktur prompt yang telah dibuat, hasil akhir yang diharapkan dari penggunaan AI adalah:

**Input Data → Pengolahan Data → Perhitungan Ts − Tp → Diagram Wadati → Regresi Linear → T₀ dan Vp/Vs → Interpretasi → Verifikasi → Kesimpulan**

Dengan struktur tersebut, AI tidak hanya memberikan jawaban akhir, tetapi juga membantu menunjukkan bagaimana suatu permasalahan geofisika dapat diselesaikan secara sistematis menggunakan kecerdasan artifisial.

### IMPLEMENTASI KODE PYTHON

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. INPUT DATA SEISMIK (Baseline waktu: 08:15:00.00 WIB)
stasiun = ['LEM', 'CBJI', 'CIS', 'TNG', 'BJI', 'CNJI']
tp_sec = np.array([28.40, 33.20, 39.50, 45.80, 53.10, 61.70]) # detik relatif
ts_sec = np.array([34.70, 43.00, 53.60, 64.60, 77.10, 92.30]) # detik relatif

# 2. PERHITUNGAN SELISIH WAKTU (Ts - Tp)
ts_minus_tp = ts_sec - tp_sec

# 3. REGRESI LINIER ORDINARY LEAST SQUARES (OLS)
# Persamaan: (Ts - Tp) = m * Tp + c
n = len(tp_sec)
sum_x = np.sum(tp_sec)
sum_y = np.sum(ts_minus_tp)
sum_xy = np.sum(tp_sec * ts_minus_tp)
sum_x2 = np.sum(tp_sec ** 2)

m = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x ** 2)
c = (sum_y - m * sum_x) / n

# 4. PENENTUAN PARAMETER GEMPA
t0 = -c / m                                           # Origin Time (s)
vp_vs = m + 1                                         # Rasio Kecepatan Vp/Vs
poisson_ratio = (vp_vs**2 - 2) / (2 * (vp_vs**2 - 1)) # Rasio Poisson

# Evaluasi Statistik (Residu, R2, RMS Error)
y_pred = m * tp_sec + c
residuals = ts_minus_tp - y_pred
ss_res = np.sum(residuals ** 2)
ss_tot = np.sum((ts_minus_tp - np.mean(ts_minus_tp)) ** 2)
r_squared = 1 - (ss_res / ss_tot)
rms_error = np.sqrt(np.mean(residuals ** 2))

# 5. CETAK HASIL KE TERMINAL
print("=" * 70)
print(f"Persamaan Regresi : (Ts - Tp) = {m:.4f} * Tp + ({c:.4f})")
print(f"Origin Time (T0)  : 08:15:{t0:05.2f} WIB  ({t0:.2f} s relatif)")
print(f"Rasio Vp/Vs       : {vp_vs:.4f}")
print(f"Rasio Poisson (v) : {poisson_ratio:.4f}")
print(f"Koefisien R²      : {r_squared:.5f} | RMS Error: {rms_error:.4f} s")
print("=" * 70)

# 6. VISUALISASI MATPLOTLIB
plt.figure(figsize=(9, 5.5), dpi=150)
x_line = np.linspace(t0, np.max(tp_sec) + 5, 200)
y_line = m * x_line + c

plt.plot(x_line, y_line, color='#1f77b4', lw=2, 
         label=f'Regresi Wadati: $(T_s - T_p) = {m:.4f} T_p {c:+.4f}$\n$R^2 = {r_squared:.5f}$')
plt.scatter(tp_sec, ts_minus_tp, color='#d62728', s=70, zorder=5, label='Data Stasiun')

for i, st in enumerate(stasiun):
    plt.annotate(f"{st}\n({tp_sec[i]:.1f}s, {ts_minus_tp[i]:.1f}s)",
                 (tp_sec[i], ts_minus_tp[i]), textcoords="offset points", 
                 xytext=(0, 10), ha='center', fontsize=8, fontweight='bold')

plt.scatter(t0, 0, color='#2ca02c', s=130, marker='*', zorder=6,
            label=f'Origin Time ($T_0$ = 08:15:{t0:05.2f})')
plt.axhline(0, color='gray', ls='--', lw=0.8)
plt.title("DIAGRAM WADATI PENENTUAN PARAMETER GEMPA BUMI", fontsize=11, fontweight='bold')
plt.xlabel("Waktu Tiba Gelombang P, $T_p$ (detik sejak 08:15:00.00)")
plt.ylabel("Selisih Waktu Tiba, $(T_s - T_p)$ (detik)")
plt.grid(True, ls=':', alpha=0.6)
plt.legend(loc='lower right')
plt.tight_layout()
plt.savefig('wadati_diagram_plot.png', dpi=300)
plt.show()
```
