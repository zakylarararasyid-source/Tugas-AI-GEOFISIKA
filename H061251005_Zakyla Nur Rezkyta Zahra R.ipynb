### GOOD PROMPT UNTUK PENYELESAIAN MASALAH GEOFISIKA

## PROMPT 1 - MENENTUKAN PARAMETER GEMPA
## 1. Narasi Permasalahan

Dalam bidang geofisika, salah satu permasalahan yang sering dihadapi oleh seorang geofisikawan adalah mengidentifikasi parameter gempa bumi berdasarkan data waktu tiba gelombang P dan gelombang S. Data tersebut dapat digunakan untuk menentukan waktu terjadinya gempa atau *origin time* serta memperkirakan parameter lain yang berkaitan dengan sumber gempa.

Salah satu metode yang dapat digunakan untuk menganalisis hubungan antara waktu tiba gelombang P dan S adalah **Diagram Wadati**. Metode ini memanfaatkan perbedaan waktu tiba gelombang S dan P dari beberapa stasiun seismik. Dengan melakukan analisis terhadap hubungan antara waktu tiba gelombang tersebut, dapat diperoleh informasi mengenai **origin time (T₀)** dan rasio kecepatan gelombang **Vp/Vs**.

Dalam praktiknya, data seismik yang diperoleh dari beberapa stasiun dapat memiliki perbedaan waktu tiba gelombang P dan S. Oleh karena itu, diperlukan proses pengolahan data yang sistematis untuk mendapatkan parameter tersebut dan memastikan hasil analisis sesuai dengan prinsip dasar seismologi.

Pada tugas ini, kecerdasan artifisial digunakan untuk membantu seorang geofisikawan dalam mengolah dan menganalisis data waktu tiba gelombang P dan S menggunakan metode Diagram Wadati. AI tidak hanya digunakan untuk memberikan hasil akhir, tetapi juga membantu menjelaskan proses pengolahan dan perhitungan secara bertahap sehingga hasil analisis dapat dipahami dan diverifikasi.

---

# 2. STRUKTUR GOOD PROMPT

Good prompt yang digunakan dalam penyelesaian masalah geofisika ini disusun berdasarkan struktur **R–T–C–I–K–O–Q–V**, yaitu:

- **R — Role:** menentukan peran AI dalam menyelesaikan masalah.
- **T — Task:** menjelaskan tugas yang harus dilakukan AI.
- **C — Context:** memberikan latar belakang permasalahan.
- **I — Input:** menentukan data yang diberikan kepada AI.
- **K — Constraints:** memberikan batasan dalam proses penyelesaian.
- **O — Output:** menentukan hasil yang diharapkan.
- **Q — Quality Criteria:** menentukan kriteria kualitas hasil.
- **V — Verification:** menentukan proses verifikasi hasil.

---

## R — ROLE

Anda berperan sebagai seorang **geofisikawan dan seismolog** yang memiliki pemahaman mengenai pengolahan data seismik, gelombang P dan S, serta metode Diagram Wadati.

Anda memiliki kemampuan dalam melakukan analisis data waktu tiba gelombang seismik, perhitungan parameter gempa, interpretasi hasil, serta melakukan pengecekan terhadap hasil perhitungan berdasarkan prinsip dasar seismologi.

---

## T — TASK

Tugas Anda adalah membantu mengolah data waktu tiba gelombang P dan S dari beberapa stasiun seismik untuk menentukan **origin time (T₀)** dan **rasio Vp/Vs** menggunakan metode Diagram Wadati.

Jelaskan proses penyelesaian secara bertahap, mulai dari pengolahan data awal, menghitung selisih waktu **Ts − Tp**, membuat hubungan Diagram Wadati, menentukan persamaan regresi, memperoleh nilai **origin time (T₀)**, menentukan **Vp/Vs**, hingga memberikan interpretasi terhadap hasil yang diperoleh.

---

## C — CONTEXT

Data yang digunakan merupakan data waktu tiba gelombang P dan S dari beberapa stasiun seismik yang merekam kejadian gempa yang sama.

Data tersebut akan dianalisis menggunakan metode Diagram Wadati untuk mengetahui hubungan antara waktu tiba gelombang P dan S. Analisis ini digunakan untuk memperkirakan **waktu asal gempa (origin time)** dan **rasio kecepatan gelombang P terhadap gelombang S (Vp/Vs)**.

Penggunaan kecerdasan artifisial dalam permasalahan ini bertujuan untuk membantu geofisikawan dalam melakukan pengolahan data secara sistematis, menjelaskan proses perhitungan, mengidentifikasi kemungkinan kesalahan, serta membantu melakukan verifikasi terhadap hasil analisis.

---

## I — INPUT

Data masukan yang diberikan kepada AI berupa tabel data seismik yang terdiri dari:

1. Nama atau kode stasiun.
2. Waktu tiba gelombang P (Tp).
3. Waktu tiba gelombang S (Ts).
4. Selisih waktu Ts − Tp.
5. Data atau parameter tambahan lain yang diperlukan dalam analisis.

AI harus menggunakan data yang diberikan oleh pengguna sebagai dasar perhitungan dan tidak diperbolehkan membuat data baru apabila data yang diperlukan belum tersedia.

Apabila terdapat data yang kurang, tidak lengkap, atau tidak konsisten, AI harus mengidentifikasi dan menjelaskan permasalahan tersebut terlebih dahulu.

---

## K — CONSTRAINTS

Dalam menyelesaikan permasalahan, AI harus mengikuti beberapa batasan berikut:

1. Menggunakan prinsip dasar seismologi dan metode Diagram Wadati.
2. Menampilkan rumus yang digunakan dalam setiap tahap perhitungan.
3. Menggunakan satuan waktu yang konsisten.
4. Tidak mengubah nilai data input tanpa alasan yang jelas.
5. Tidak membuat data baru untuk menggantikan data yang tidak tersedia.
6. Menjelaskan proses perhitungan sebelum memberikan hasil akhir.
7. Menggunakan bahasa Indonesia yang jelas dan mudah dipahami oleh mahasiswa geofisika.
8. Membedakan antara data awal, hasil perhitungan, dan hasil interpretasi.
9. Jika terdapat ketidakpastian atau kesalahan pada data, AI harus menyampaikannya secara jelas.

---

## O — OUTPUT

AI diharapkan menghasilkan keluaran berupa analisis data seismik menggunakan metode Diagram Wadati secara sistematis.

Output yang dihasilkan meliputi:

1. **Tabel hasil pengolahan data**  
   Menampilkan data stasiun, waktu tiba gelombang P (Tp), waktu tiba gelombang S (Ts), dan selisih waktu Ts − Tp.

2. **Proses perhitungan**  
   Menjelaskan langkah-langkah perhitungan secara bertahap beserta rumus yang digunakan.

3. **Diagram Wadati**  
   Menampilkan hubungan antara waktu tiba gelombang P dan selisih waktu Ts − Tp dalam bentuk grafik.

4. **Persamaan regresi linear**  
   Menampilkan persamaan garis hasil analisis serta nilai kemiringan (*slope*) dan intersep.

5. **Parameter gempa**  
   Menentukan:
   - Origin time (T₀)
   - Rasio Vp/Vs
   - Parameter lain yang dapat diperoleh dari data apabila tersedia.

6. **Interpretasi hasil**  
   Menjelaskan arti nilai T₀ dan Vp/Vs dalam konteks kondisi dan karakteristik gelombang seismik.

7. **Verifikasi hasil**  
   Melakukan pengecekan ulang terhadap perhitungan dan memastikan hasil yang diperoleh konsisten dengan data awal.

8. **Kesimpulan**  
   Memberikan kesimpulan singkat mengenai hasil analisis dan parameter gempa yang diperoleh.

---

## Q — QUALITY CRITERIA

Hasil analisis yang diberikan harus memenuhi kriteria berikut:

- Benar secara konsep dan prinsip dasar seismologi.
- Memiliki perhitungan yang konsisten.
- Menggunakan rumus yang sesuai dengan metode Diagram Wadati.
- Menjelaskan setiap tahapan secara logis dan sistematis.
- Mencantumkan satuan pada hasil perhitungan.
- Mudah dipahami oleh mahasiswa geofisika.
- Membedakan antara data, hasil perhitungan, dan interpretasi.
- Tidak memberikan kesimpulan yang tidak didukung oleh data.
- Hasil akhir harus dapat ditelusuri kembali berdasarkan data dan proses perhitungan yang digunakan.

---

## V — VERIFICATION

Setelah memperoleh hasil akhir, AI harus melakukan proses verifikasi dengan cara:

1. Memeriksa kembali perhitungan Ts − Tp.
2. Memeriksa konsistensi satuan waktu.
3. Memeriksa apakah nilai T₀ secara logis berada sebelum waktu tiba gelombang P.
4. Memeriksa hubungan antara kemiringan (*slope*) Diagram Wadati dengan nilai Vp/Vs.
5. Memeriksa kembali persamaan regresi yang diperoleh.
6. Menggunakan persamaan yang diperoleh untuk melakukan pengecekan terhadap data awal.
7. Membandingkan hasil perhitungan dengan prinsip dasar seismologi.
8. Jika ditemukan ketidaksesuaian, menjelaskan kemungkinan sumber kesalahan dan melakukan koreksi berdasarkan data yang tersedia.

# PROMPT 3 — INTERPRETASI DATA GEOLISTRIK RESISTIVITAS

## 1. Narasi Permasalahan

Metode geolistrik resistivitas merupakan salah satu metode geofisika yang digunakan untuk mengetahui variasi sifat kelistrikan material bawah permukaan. Metode ini dapat digunakan untuk membantu mengidentifikasi lapisan tanah dan batuan, zona jenuh air, serta kemungkinan keberadaan struktur bawah permukaan tertentu.

Dalam pengolahan data geolistrik, seorang geofisikawan perlu menganalisis nilai resistivitas berdasarkan data hasil pengukuran. Salah satu tantangannya adalah menghubungkan nilai resistivitas dengan kemungkinan jenis material penyusun bawah permukaan.

Kecerdasan artifisial dapat digunakan untuk membantu mengolah data resistivitas, mengelompokkan nilai resistivitas, dan memberikan interpretasi awal berdasarkan nilai resistivitas serta konteks geologi daerah penelitian.

---

## 2. Struktur Good Prompt

### R — Role

Anda berperan sebagai seorang **geofisikawan yang ahli dalam metode geolistrik resistivitas dan interpretasi kondisi bawah permukaan**.

### T — Task

Tugas Anda adalah membantu menganalisis data resistivitas bawah permukaan dan memberikan interpretasi terhadap kemungkinan jenis material atau lapisan berdasarkan nilai resistivitas yang diperoleh.

### C — Context

Data geolistrik diperoleh melalui pengukuran sifat kelistrikan bawah permukaan menggunakan konfigurasi elektroda tertentu.

Nilai resistivitas dapat memberikan informasi mengenai karakteristik material bawah permukaan. Namun, interpretasi harus mempertimbangkan kondisi geologi dan tidak hanya berdasarkan nilai resistivitas secara tunggal.

### I — Input

Data yang diberikan berupa:

1. Jarak atau konfigurasi elektroda.
2. Nilai arus listrik.
3. Nilai beda potensial.
4. Nilai resistivitas semu.
5. Data geologi pendukung apabila tersedia.

### K — Constraints

1. Gunakan prinsip dasar metode geolistrik resistivitas.
2. Tampilkan rumus apabila diperlukan.
3. Jangan membuat data yang tidak diberikan.
4. Jangan langsung menyatakan jenis batuan hanya berdasarkan satu nilai resistivitas.
5. Pertimbangkan konteks geologi.
6. Jelaskan ketidakpastian interpretasi.
7. Gunakan bahasa Indonesia yang mudah dipahami mahasiswa.

### O — Output

Hasilkan:

1. Tabel pengolahan data.
2. Perhitungan resistivitas apabila diperlukan.
3. Grafik hubungan parameter pengukuran dengan resistivitas.
4. Identifikasi pola atau zona resistivitas.
5. Interpretasi kemungkinan material bawah permukaan.
6. Hubungan hasil resistivitas dengan kondisi geologi.
7. Tingkat keyakinan atau keterbatasan interpretasi.
8. Kesimpulan.

### Q — Quality Criteria

Hasil harus:

- berdasarkan prinsip geolistrik;
- menggunakan perhitungan yang benar;
- tidak melakukan interpretasi secara berlebihan;
- mempertimbangkan data geologi;
- menjelaskan hubungan antara nilai resistivitas dan material bawah permukaan;
- mudah dipahami mahasiswa geofisika.

### V — Verification

Lakukan verifikasi dengan:

1. Memeriksa kembali perhitungan resistivitas.
2. Memeriksa satuan.
3. Membandingkan pola resistivitas dengan data pendukung.
4. Memeriksa apakah interpretasi sesuai dengan kondisi geologi.
5. Menjelaskan kemungkinan penyebab ketidakpastian hasil.
