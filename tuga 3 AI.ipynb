## 2. Latar Belakang

Data waktu tiba gelombang seismik merupakan salah satu informasi penting dalam analisis gempa bumi. Gelombang P (*Primary*) dan gelombang S (*Secondary*) memiliki kecepatan rambat yang berbeda sehingga tiba di stasiun seismik pada waktu yang berbeda.

Metode Wadati dapat digunakan untuk menganalisis hubungan antara waktu tiba gelombang P dan selisih waktu tiba gelombang S terhadap gelombang P. Dengan menerapkan regresi linear, hubungan tersebut dapat digunakan untuk memperkirakan rasio kecepatan gelombang P terhadap gelombang S (\(V_P/V_S\)).

Recipe ini dirancang untuk membantu proses analisis data gempa secara sistematis menggunakan Python, mulai dari membaca data, memeriksa kualitas data, melakukan regresi linear, hingga menghasilkan grafik dan laporan analisis.

## 3. Tujuan

1. Membaca dan memeriksa data waktu tiba gelombang P dan S.
2. Menghitung selisih waktu tiba gelombang S dan P (\(T_S-T_P\)).
3. Menerapkan regresi linear pada data Metode Wadati.
4. Memperkirakan rasio kecepatan gelombang P dan S.
5. Menghasilkan grafik dan laporan yang dapat diperiksa serta diulang.

## 4. Dasar Teori

Gelombang P merupakan gelombang badan yang umumnya memiliki kecepatan rambat lebih tinggi daripada gelombang S. Oleh karena itu, gelombang P biasanya tiba lebih dahulu di stasiun seismik.

Selisih waktu tiba kedua gelombang dapat dituliskan sebagai:

\[
T_S-T_P
\]

Hubungan yang digunakan dalam analisis Wadati adalah:

\[
T_S-T_P=
\left(\frac{V_P}{V_S}-1\right)(T_P-T_0)
\]

Persamaan regresi linear dinyatakan sebagai:

\[
y=ax+b
\]

dengan:

- \(y=T_S-T_P\)
- \(x=T_P\)
- \(a\) = kemiringan garis regresi (*slope*)
- \(b\) = intersep garis regresi
- \(T_0\) = waktu asal gempa (*origin time*)

Dengan asumsi dan acuan waktu yang sesuai, rasio kecepatan gelombang P terhadap gelombang S dapat diperkirakan melalui:

\[
\frac{V_P}{V_S}=a+1
\]

Interpretasi waktu asal gempa dari intersep perlu memperhatikan acuan waktu yang digunakan dalam data. Data waktu tiba harus konsisten sebelum hasilnya ditafsirkan secara geofisika.
